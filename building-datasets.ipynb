```python
import os
import json

# Path to the directory containing JSON files
directory_path = "deep-saved-logs"

# Function to check if a file should be deleted based on its content
def should_delete(file_data):
    # Replace with your actual condition logic
    # Example condition: delete if the "type" field equals "delete"
    return ',' in file_data["info"]["map"]

# Iterate through files in the directory
for entry in os.scandir(directory_path):
    if entry.is_file() and entry.name.endswith(".json"):
        file_path = entry.path
        
        # Open and read the JSON file
        with open(file_path, 'r') as file:
            try:
                file_data = json.load(file)
            except json.JSONDecodeError:
                print(f"Error decoding JSON in file {file_path}. Skipping.")
                continue
        
        # Check the condition and delete the file if needed
        if should_delete(file_data):
            try:
                # os.remove(file_path)
                print(f"Deleted file: {file_path}")
            except Exception as e:
                print(f"Error deleting file {file_path}: {e}")

print("Processing complete.")
```
